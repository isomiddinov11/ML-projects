```python
from scipy.stats import chi2_contingency
alpha = 0.05
```

```python
df_cleaned['price_bin'] = pd.qcut(df_cleaned['price'], q=4, labels=False)

categorical_cols = df_cleaned.select_dtypes(include=['object', 'category']).columns.tolist()
binary_cols = ['accident_history', 'owners', 'city_tier', 'fuel_type', 'transmission'] 
cat_features = list(set(categorical_cols + [col for col in binary_cols if col in df_cleaned.columns]))

results = []

for col in cat_features:
    contingency_table = pd.crosstab(df_cleaned[col], df_cleaned['price_bin'])
    chi2, p, dof, ex = chi2_contingency(contingency_table)
    
    decision = "Saqlash" if p < alpha else "Tashlash"
    
    results.append({
        'Feature': col,
        'Chi2 Statistic': round(chi2, 4),
        'p-value': round(p, 5),
        'Qaror': decision
    })

chi2_df = pd.DataFrame(results)
chi2_df
```
