```python
from sklearn.cluster import KMeans
import numpy as np

mijozlar_malumoti = np.array([
    [200, 3], [220, 4], [50, 1],
    [60, 2], [800, 10], [850, 12]
])

klaster_soni = 3
model = KMeans(n_clusters=klaster_soni, random_state=42)
model.fit(mijozlar_malumoti)

guruhlar = model.labels_
markazlar = model.cluster_centers_

print("mijoz tushgan har bir guruh: ", guruhlar)
```

```python
from sklearn.decomposition import PCA
xususiyatlar = np.array([
    [5,90,3,8], [4,85,2,7],
    [2,40,8,3], [1,30,9,2]
])
pca = PCA(n_components=2)
siqilgan_malumot = pca.fit_transform(xususiyatlar)
print("2 olchami siqilgan malumot: ", siqilgan_malumot)
```

```python
from sklearn.decomposition import PCA
xususiyat = np.array([
    [90,95,8,9], [85,90,7,8],
    [40,50,3,2], [35,45,2,3]
    ])
pca = PCA(n_components=1)
siqilgan_malumot = pca.fit_transform(xususiyat)
print("2 olchami siqilgan malumot: ", siqilgan_malumot.ravel())
```

```python
from sklearn.ensemble import IsolationForest
tranzaksiyalar = np.array([[45000], [52000], [48000], [51000],
                           [49000], [47000], [500000]])

model = IsolationForest(contamination = 0.15,
                        random_state=42)
natija = model.fit_predict(tranzaksiyalar)
print("har bir otkazma holati: ", natija)
```
