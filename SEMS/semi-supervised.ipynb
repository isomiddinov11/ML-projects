```python
import numpy as np
import pandas as pd
```

```python
malumot = pd.DataFrame({
    'xarid_miqdori':[10.850,20,900,15,780,30,850,25,800],
    'oxirgi_xarid_kun':[300,5,280,3,320,8,290,6,310],  
    'yorliq':[0,-1,-1,-1,-1,-1,-1,-1,-1]
})

print(malumot)
```

```python
from sklearn.preprocessing import StandardScaler

X = malumot[['xarid_miqdori','oxirgi_xarid_kun']]  
y = malumot['yorliq']

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
print("xususiyatlar:\n", X_scaled)
print("yorliqlar:", y.values)
```

```python
from sklearn.semi_supervised import LabelSpreading

model = LabelSpreading(kernel='knn', n_neighbors=2)
model.fit(X_scaled, y)

malumot = pd.DataFrame()

malumot['bashorat_yorliq'] = model.transduction_
malumot['ishonch'] = model.label_distributions_.max(axis=1).round(2)

print(malumot)
```
