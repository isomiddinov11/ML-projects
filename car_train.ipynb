```python
def mashina_narxini_bashorat_qil(car_age, engine_size, has_accident, mileage, model, ustunlar):
    yangi_mashina = pd.DataFrame([{
        'car_age': car_age,
        'engine_size': engine_size,
        'has_accident': has_accident,
        'mileage': mileage
    }])

    yangi_mashina_encoded = pd.get_dummies(yangi_mashina, columns=['has_accident'], dtype=int)

    yangi_mashina_encoded = yangi_mashina_encoded.reindex(columns=ustunlar, fill_value=0)

    natija = model.predict(yangi_mashina_encoded)
    return natija[0]


ustunlar_tartibi = eng_yaxshi_model.feature_names_in_

misol_narx = mashina_narxini_bashorat_qil(
    car_age=5,
    engine_size=1.5,
    has_accident='no',
    mileage=45000,
    model=eng_yaxshi_model,
    ustunlar=ustunlar_tartibi
)

print(f"Taxmin qilingan mashina narxi: ${misol_narx:,.2f}")
```
